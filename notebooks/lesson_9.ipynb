## Bayes' Rule

Bayes’ rule is a way of **updating a belief when we see new evidence**.

$$
P(A \mid B) = \frac{P(B \mid A)\cdot P(A)}{P(B)}
$$


### What do the symbols mean?

* $A$ and $B$ are just **events** (things that can happen)
* $P(A)$ means: the probability that $A$ happens
* $P(A \mid B)$ means: the probability that $A$ happens **given that** we know $B$ has happened
* The symbol $\mid$ means **“given that”**

So if:

* $A$ = “the plane crashes”
* $B$ = “the plane shakes”

then:

* $P(A)$ = the probability that a plane crashes
* $P(A \mid B)$ = the probability that a plane crashes **given that it is shaking**


### The parts of Bayes' rule

* $P(A)$: the **prior** — your starting belief  
* $P(B \mid A)$: the **likelihood** — how well $A$ predicts the evidence  
* $P(B)$: the **base rate of the evidence** — how common $B$ is overall  
* $P(A \mid B)$: the **posterior** — your updated belief  


### Bayes' rule in human language (as a fraction)

We can rewrite the formula directly in words:

$$
\text{Updated belief} =
\frac{
\text{(how well the hypothesis predicts the evidence)} \times \text{(how common the hypothesis was to begin with)}
}{
\text{how common the evidence is overall}
}
$$

Or more compactly:

$$
\text{Posterior} =
\frac{
\text{Likelihood} \times \text{Prior}
}{
\text{Evidence frequency}
}
$$


### Intuition

Bayes’ rule is a disciplined way of reasoning:

> Start with a prior belief.  
> Then update it when new evidence appears.

The key idea:

> **Evidence only matters if it is rare.**  
> If the evidence is common anyway, it tells us very little.


## Example: Turbulence and Plane Crashes

Let:

* $A$ = the plane **crashes**  
* $B$ = the plane **has turbulence**  

We observe turbulence and want:

$$
P(A \mid B) = P(\text{crash} \mid \text{turbulence})
$$


### Step 1: Start with the prior

$$
P(A) = \frac{51}{38{,}700{,}000} \approx 0.0000013
$$

Crashes are extremely rare.


### Step 2: Likelihood

$$
P(B \mid A) \approx 1
$$

If a plane crashes, it almost certainly involves shaking.


### Step 3: Evidence frequency

$$
P(B) = 0.01
$$

Turbulence is relatively common.


### Step 4: Update

$$
P(A \mid B) = \frac{1 \cdot 0.0000013}{0.01} = 0.00013
$$

$$
0.00013 = 0.013\%
$$

## A second example: reasoning about people

We can also use Bayes’ rule when we **do not have exact data**, and instead rely on our own judgments.

Let:

* $A$ = the student is **generally disengaged / not taking the course seriously**
* $B$ = the student **speaks negatively about the course**

We overhear a student criticizing the course, and we ask:

$$
P(A \mid B) = P(\text{disengaged} \mid \text{negative comment})
$$

---

### Step 1: Start with a prior

Before hearing anything, we might think:

> Most students are reasonably engaged.

So we might set:

$$
P(A) \approx 0.2
$$

(20% of students are disengaged — this is a subjective judgment)

---

### Step 2: Likelihood

How likely is it that a disengaged student speaks negatively?

Probably quite likely:

$$
P(B \mid A) \approx 0.7
$$

---

### Step 3: How common is the evidence?

Here is the key question:

> How often do *any* students complain or say something negative?

Even engaged students sometimes complain — about difficulty, timing, mood, etc.

So maybe:

$$
P(B) \approx 0.4
$$

---

### Step 4: Update

$$
P(A \mid B) = \frac{0.7 \cdot 0.2}{0.4} = 0.35
$$

---

### Interpretation

After hearing the negative comment:

* The probability the student is disengaged increases
* But it is still far from certain

---

### The key insight

> Saying something negative is **not rare behavior**.

Even good, engaged students sometimes say:

* “this is boring”
* “this is hard”
* “this is annoying”

So:

> **The evidence (complaining) is weak, because it is common.**

---

### Final lesson

Bayes’ rule applies even when we do not have perfect data.

We are always implicitly using:

* assumptions about how common things are (priors)
* assumptions about behavior (likelihoods)

The mistake people make is this:

> They treat a single piece of evidence as decisive,
> without asking how common that behavior is.

---

### Blunt takeaway

> **Hearing one negative comment is not strong evidence that a student is disengaged —**
> because even engaged students complain sometimes.

```python
# Step 1: Take the original release-decision column
# We turn it into text ("string") and remove extra spaces
decision = df["Release Decision at Arraign"].astype("string").str.strip()

# Step 2: Create a dictionary that translates legal decision categories into numbers
# 1 = released
# 0 = not released / detained
release_map = {
    "ROR": 1,                  # released on recognizance
    "Nonmonetary release": 1,  # released, but with conditions
    "Bail-set": 0,             # not simply released
    "Remanded": 0              # kept in custody
}

# Step 3: Use the dictionary to create a new cleaned column
# This gives us a simpler version of the release decision
df["released_clean"] = decision.map(release_map)

# Step 4: Keep only the rows where the release decision is clear
# If a value was not in our dictionary, it becomes missing (NaN), and we drop it here
analysis_df = df[df["released_clean"].notna()].copy()

# Step 5: Create a second outcome variable called "detained"
# If released_clean = 1, then detained = 0
# If released_clean = 0, then detained = 1
analysis_df["detained"] = 1 - analysis_df["released_clean"]

# Step 6: Show the proportion of cases that were detained vs released
# normalize=True means we get percentages rather than raw counts
analysis_df["detained"].value_counts(normalize=True)
```

```python
# Quick check: how many cases fall into each original category?
decision.value_counts(dropna=False) ## these are the ones we turn into binary data
```

```python
# Step 7: Compute the baseline probability (the PRIOR)

prior = analysis_df["detained"].mean()

print("Baseline probability of detention (PRIOR):")
print(round(prior, 3))
```

## The prior

This number is our starting point.

> If we know nothing about a defendant, this is our best estimate
> of the probability that they will be detained.

In Bayesian terms:

> This is the **prior**.


```python
# Step 8: Define one piece of evidence

# Example: "many prior misdemeanors"
analysis_df["many_priors"] = analysis_df["prior_misd_cnt"] >= 3

analysis_df["many_priors"].value_counts()
```

```python
# Step 9: Compute the updated probability (POSTERIOR)

posterior = analysis_df.loc[
    analysis_df["many_priors"], "detained"
].mean()

print("Probability of detention given many prior misdemeanors (POSTERIOR):")
print(round(posterior, 3))
```

## What just happened?

We started with a baseline probability:

* How often are people detained overall?

This is:

$$
P(\text{detained})
$$

Then we added new information:

* This person has many prior offenses

And we recomputed the probability:

$$
P(\text{detained} \mid \text{many priors})
$$


## This is Bayesian reasoning

The general formula is:

$$
P(A \mid B) = \frac{P(B \mid A)\cdot P(A)}{P(B)}
$$

In our example:

* $A$ = “the person is detained”
* $B$ = “the person has many prior offenses”

So the quantity we care about is:

$$
P(\text{detained} \mid \text{many priors})
$$

> the probability that someone is detained **given that** they have many prior offenses


## The parts of the formula

* **Prior** = $P(\text{detained})$
  the overall detention rate (before we know anything)

* **Evidence** = $B = \text{many priors}$
  the new information we observe

* **Posterior** = $P(\text{detained} \mid \text{many priors})$
  the updated probability after seeing that information


## What changed?

We moved from:

$$
P(\text{detained})
$$

to:

$$
P(\text{detained} \mid \text{many priors})
$$

So we started with a general belief, and then **updated it after observing evidence**.


## Important clarification

We did not explicitly calculate the full formula.

Instead, we used data to directly compute the conditional probability:

$$
P(\text{detained} \mid \text{many priors})
$$

This gives us the same quantity that Bayes’ rule defines.


## The key question

> How much did the probability change?

That tells us how informative the evidence is.



```python
#Baseline probability of detention (PRIOR):
base_rate_prob_of_detention = 0.198

prob_of_detention_given_many_priors = 0.342

print(f'You increase your likelihood of detention by {round(prob_of_detention_given_many_priors - base_rate_prob_of_detention, 2)} if you have 3 or more prior convictions.')
```

## From updating beliefs to testing differences

So far, we updated a probability:

* from $P(\text{detained})$
* to $P(\text{detained} \mid \text{many priors})$

We saw that the probability changed when we added new information.

> Could it just be random noise?

```python
from scipy.stats import chi2_contingency

# Build contingency table
gender_table = pd.crosstab(analysis_df["Gender"], analysis_df["detained"])

print("Contingency table:")
print(gender_table)

# Chi-square test
chi2_gender, p_gender, dof_gender, expected_gender = chi2_contingency(gender_table)

print("\nChi-square:", round(chi2_gender, 2))
print("p-value:", p_gender)
```

## What does the p-value tell us?

The p-value answers this question:

> If there were actually no real difference between men and women,
> how likely is it that we would observe a difference this large just by chance?


### Interpretation

* **Small p-value** → unlikely to be random → difference is likely real  
* **Large p-value** → could be random → difference may not be meaningful  


### Important

Statistical significance does NOT mean:

* the difference is large  
* the difference is fair  
* the difference is justified  

It only means:

> The difference is unlikely to be due to chance.

## But there is a problem

With large datasets, even very small differences become statistically significant.

So we now ask a second question:

> Even if the difference is real, is it actually large?

To answer this, we measure **effect size**.

## Measuring effect size

We use **Cramér’s V** to measure effect size.

This tells us:

> How strong is the relationship between group and outcome?


## Interpretation (rule of thumb)

* 0.00 → no relationship  
* ~0.10 → small  
* ~0.30 → moderate  
* ~0.50 → strong  


## Important

> A result can be statistically significant but still have a small effect size.

So now we measure:

> How large is the difference we found?

```python
import numpy as np

# Function to calculate Cramér's V
def cramers_v(chi2, table):
    n = table.sum().sum()
    r, k = table.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))
```

```python
# Effect size for gender

cramers_v_gender = cramers_v(chi2_gender, gender_table)

print("Cramér's V (gender):", round(cramers_v_gender, 3))
```

```python
## Interpreting the results

We found:

* Gender: Cramér’s V ≈ 0.118

---

## What does this mean?

> The differences are real — but small.

---

## Key lesson

With large datasets:

* It is easy to detect small differences
* These differences can be statistically significant

But:

> Small differences are not necessarily important.

---

## Takeaway

* p-value → tells us if a difference exists
* effect size → tells us how large it is

You need both.
```

```python
# =========================================================
# Machine learning model!
# =========================================================

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix

# Step 1: Choose features
features = [
    "Age_at_Arrest",
    "prior_misd_cnt",
    "prior_nonvfo_cnt",
    "prior_vfo_cnt",
    "pend_misd",
    "pend_nonvfo",
    "pend_vfo",
    "supervision",
    "Gender"   # one simple categorical variable
]

target = "rearrested"

ml_df = analysis_df[features + [target, "detained"]].copy()

# Step 2: Drop missing values
ml_df = ml_df.dropna().copy()

print("Rows used for model:", ml_df.shape[0])

# Step 3: Convert Gender to numeric (keep it simple)
ml_df["Gender"] = ml_df["Gender"].map({
    "Male": 1,
    "Female": 0
})

# Drop rows where gender was unknown
ml_df = ml_df.dropna().copy()

# Step 4: Define X and y
X = ml_df[features]
y = ml_df[target]
judge_signal = ml_df["detained"]

# Step 5: Train/test split
X_train, X_test, y_train, y_test, judge_train, judge_test = train_test_split(
    X, y, judge_signal, test_size=0.3, random_state=42
)

# Step 6: Train model
model = LogisticRegression(max_iter=2000)
model.fit(X_train, y_train)

# Step 7: Predictions
model_probs = model.predict_proba(X_test)[:, 1]
model_preds = (model_probs >= 0.5).astype(int)

# Step 8: Compare model vs judges
model_auc = roc_auc_score(y_test, model_probs)
judge_auc = roc_auc_score(y_test, judge_test)

model_acc = accuracy_score(y_test, model_preds)
judge_acc = accuracy_score(y_test, judge_test)

print("\nMODEL")
print("AUC:", round(model_auc, 3))
print("Accuracy:", round(model_acc, 3))

print("\nJUDGES")
print("AUC:", round(judge_auc, 3))
print("Accuracy:", round(judge_acc, 3))

# Step 9: Confusion matrices
print("\nModel confusion matrix:")
print(confusion_matrix(y_test, model_preds))

print("\nJudge confusion matrix:")
print(confusion_matrix(y_test, judge_test))

# Step 10: Coefficients
coef_table = pd.DataFrame({
    "feature": features,
    "coefficient": model.coef_[0]
}).sort_values("coefficient", ascending=False)

print("\nModel coefficients:")
print(coef_table)
```

## 1. What is accuracy?

**Accuracy** is the share of cases where a prediction is correct.

In fraction form:

$$
\text{Accuracy} = \frac{\text{correct predictions}}{\text{all predictions}}
$$

So if we correctly classify 80 out of 100 cases, accuracy is:

$$
\frac{80}{100} = 0.80 = 80\%
$$

In our results:

* **Model accuracy = 0.796**
* **Judge accuracy = 0.645**

So the model is correct about **79.6%** of the time, while the judges' decisions line up with the outcome about **64.5%** of the time.

But there is a catch.

### Why accuracy can mislead us

Most people in this dataset are **not rearrested**.

That means a very dumb rule like:

> "predict no rearrest for everyone"

would already get many cases right.

So a high accuracy does **not** automatically mean a method is truly good at identifying risk.

Accuracy tells us:

> "How often was this method right overall?"

But it does **not** tell us well:

> "How well does this method separate high-risk from low-risk people?"

That is why we also use **AUC**.


## 2. What is AUC?

**AUC** stands for **Area Under the Curve**, specifically the **ROC curve**.

That sounds technical, but the basic idea is simple:

> AUC measures how well a method ranks risk.

A good way to think about it is this:

> If we randomly pick one person who **was rearrested** and one person who **was not rearrested**, AUC is the probability that the method gives the higher-risk score to the person who was rearrested.

So:

* **AUC = 0.50** means the method is basically random
* **AUC = 1.00** means perfect separation
* **AUC below 0.50** means the ranking is going in the wrong direction

In our results:

* **Model AUC = 0.686**
* **Judges AUC = 0.449**

So the model has a **68.6% chance** of ranking a rearrested person above a non-rearrested person.

That is not amazing, but it is clearly better than random.

The judges' detention decision, as used here, gives an AUC of **0.449**, which means:

> detention is a very weak signal of later rearrest in this dataset.

In other words:

> the judges' yes/no detention choice does not sort people into higher and lower rearrest risk very well.


## 3. A Bayesian way to understand this

Bayesian reasoning is about updating beliefs.

We begin with some **base rate**:

$$
P(\text{rearrest})
$$

Then we use information to update that belief:

$$
P(\text{rearrest} \mid \text{features})
$$

A good predictive model is useful because it helps us move from a general prior belief to a more informed posterior belief.

So instead of saying:

> "some people get rearrested"

we say:

> "given this person's age, prior record, pending charges, and other factors, the probability changes."

That is the core Bayesian idea:

> evidence updates probability.

The model is doing this systematically.  
The judges are also making judgments under uncertainty, but their detention decision is not giving us as strong a signal of rearrest as the model does.
