```python
import matplotlib.pyplot as plt
from collections import defaultdict

# =========================================================
# QUIZ DATA
# =========================================================
# question_type:
# - "tf" = True/False
# - "choice" = one of the listed options

questions = [
    # -------------------------
    # Round 1
    # -------------------------
    {
        "round": 1,
        "question": "1. The elephant is the world's largest mammal.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 1,
        "question": "2. Sea otters sometimes hold hands while they sleep.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 1,
        "question": "3. Centipedes have more legs than any other animal.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 1,
        "question": "4. Mammals and dinosaurs coexisted.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 1,
        "question": "5. Bears can't climb trees.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 1,
        "question": "6. Camels store water in their humps.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 1,
        "question": "7. Flamingos are pink because they eat shrimp.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 1,
        "question": "8. The giant panda eats mostly bamboo.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 1,
        "question": "9. The platypus is the only mammal that lays eggs.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 1,
        "question": "10. A mule is a cross between a male donkey and a female horse.",
        "type": "tf",
        "answer": True
    },

    # -------------------------
    # Round 2
    # -------------------------
    {
        "round": 2,
        "question": "11. Who was born first?",
        "type": "choice",
        "options": ["Julius Caesar", "Confucius"],
        "answer": "Confucius"
    },
    {
        "round": 2,
        "question": "12. Who was born first?",
        "type": "choice",
        "options": ["Fidel Castro", "Mahatma Gandhi"],
        "answer": "Mahatma Gandhi"
    },
    {
        "round": 2,
        "question": "13. Who was born first?",
        "type": "choice",
        "options": ["Nelson Mandela", "Anne Frank"],
        "answer": "Nelson Mandela"
    },
    {
        "round": 2,
        "question": "14. Who was born first?",
        "type": "choice",
        "options": ["Cleopatra", "Muhammad"],
        "answer": "Cleopatra"
    },
    {
        "round": 2,
        "question": "15. Who was born first?",
        "type": "choice",
        "options": ["William Shakespeare", "Joan of Arc"],
        "answer": "Joan of Arc"
    },
    {
        "round": 2,
        "question": "16. Who was born first?",
        "type": "choice",
        "options": ["George Washington", "Sun Tzu"],
        "answer": "Sun Tzu"
    },
    {
        "round": 2,
        "question": "17. Who was born first?",
        "type": "choice",
        "options": ["Genghis Khan", "Leonardo da Vinci"],
        "answer": "Genghis Khan"
    },
    {
        "round": 2,
        "question": "18. Who was born first?",
        "type": "choice",
        "options": ["Queen Victoria", "Karl Marx"],
        "answer": "Karl Marx"
    },
    {
        "round": 2,
        "question": "19. Who was born first?",
        "type": "choice",
        "options": ["Saddam Hussein", "Marilyn Monroe"],
        "answer": "Marilyn Monroe"
    },
    {
        "round": 2,
        "question": "20. Who was born first?",
        "type": "choice",
        "options": ["Albert Einstein", "Mao Zedong"],
        "answer": "Albert Einstein"
    },

    # -------------------------
    # Round 3
    # -------------------------
    {
        "round": 3,
        "question": "21. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Germany", "France"],
        "answer": "Germany"
    },
    {
        "round": 3,
        "question": "22. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Japan", "South Korea"],
        "answer": "Japan"
    },
    {
        "round": 3,
        "question": "23. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Brazil", "Argentina"],
        "answer": "Brazil"
    },
    {
        "round": 3,
        "question": "24. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Egypt", "Botswana"],
        "answer": "Egypt"
    },
    {
        "round": 3,
        "question": "25. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Mexico", "Guatemala"],
        "answer": "Mexico"
    },
    {
        "round": 3,
        "question": "26. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Panama", "Belize"],
        "answer": "Panama"
    },
    {
        "round": 3,
        "question": "27. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Jamaica", "Haiti"],
        "answer": "Haiti"
    },
    {
        "round": 3,
        "question": "28. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Greece", "Norway"],
        "answer": "Greece"
    },
    {
        "round": 3,
        "question": "29. Which country had more people in 2026?",
        "type": "choice",
        "options": ["China", "India"],
        "answer": "India"
    },
    {
        "round": 3,
        "question": "30. Which country had more people in 2026?",
        "type": "choice",
        "options": ["Iraq", "Iran"],
        "answer": "Iran"
    },

    # -------------------------
    # Round 4
    # -------------------------
    {
        "round": 4,
        "question": "31. Mars has one moon, just like Earth.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 4,
        "question": "32. Scurvy is caused by a deficit of vitamin C.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 4,
        "question": "33. Brass is made from iron and copper.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 4,
        "question": "34. One tablespoon of oil has more calories than one tablespoon of butter.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 4,
        "question": "35. Helium is the lightest element.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 4,
        "question": "36. The common cold is caused by bacteria.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 4,
        "question": "37. The deepest place on Earth is in the Pacific Ocean.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 4,
        "question": "38. Seasons are caused by the Earth orbiting the Sun in an elliptical path.",
        "type": "tf",
        "answer": False
    },
    {
        "round": 4,
        "question": "39. Jupiter is the largest planet in our solar system.",
        "type": "tf",
        "answer": True
    },
    {
        "round": 4,
        "question": "40. The atoms in a solid are more densely packed than the atoms in a gas.",
        "type": "tf",
        "answer": True
    },
]

ALLOWED_CONFIDENCE = [55, 65, 75, 85, 95]

# =========================================================
# HELPERS
# =========================================================
def parse_tf(user_input):
    text = user_input.strip().lower()
    if text in ["t", "true"]:
        return True
    elif text in ["f", "false"]:
        return False
    else:
        return None

def parse_choice(user_input, options):
    text = user_input.strip().lower()

    # Allow full answer text
    for opt in options:
        if text == opt.lower():
            return opt

    # Allow 1 / 2 input
    if text in ["1", "2"]:
        return options[int(text) - 1]

    return None

def get_confidence():
    while True:
        try:
            c = int(input("Confidence? Choose one of 55, 65, 75, 85, 95: "))
            if c in ALLOWED_CONFIDENCE:
                return c
            print("Please enter exactly one of: 55, 65, 75, 85, 95")
        except:
            print("Enter a number.")

# =========================================================
# RUN QUIZ
# =========================================================
results = []

print("Welcome to the calibration quiz.")
print("Answer each question, then state how confident you are.\n")

current_round = None

for q in questions:
    if q["round"] != current_round:
        current_round = q["round"]
        print("\n" + "="*50)
        print(f"ROUND {current_round}")
        print("="*50)

    print("\n" + q["question"])

    if q["type"] == "tf":
        while True:
            ans = input("Your answer (T/F): ")
            parsed = parse_tf(ans)
            if parsed is not None:
                break
            print("Please type T, F, True, or False.")

    elif q["type"] == "choice":
        print("Options:")
        for i, opt in enumerate(q["options"], start=1):
            print(f"  {i}. {opt}")
        while True:
            ans = input("Your answer (type 1/2 or the full name): ")
            parsed = parse_choice(ans, q["options"])
            if parsed is not None:
                break
            print("Please type 1, 2, or one of the option names exactly.")

    confidence = get_confidence()
    correct = parsed == q["answer"]

    results.append({
        "question": q["question"],
        "your_answer": parsed,
        "correct_answer": q["answer"],
        "confidence": confidence,
        "correct": correct
    })

    if correct:
        print("Correct.")
    else:
        print(f"Wrong. Correct answer: {q['answer']}")

# =========================================================
# SUMMARY
# =========================================================
total_correct = sum(r["correct"] for r in results)
n = len(results)
overall_accuracy = total_correct / n

print("\n" + "="*50)
print("FINAL RESULTS")
print("="*50)
print(f"Score: {total_correct}/{n}")
print(f"Overall accuracy: {overall_accuracy:.1%}")

# =========================================================
# CALIBRATION DATA
# =========================================================
bucket_counts = defaultdict(int)
bucket_correct = defaultdict(int)

for r in results:
    c = r["confidence"]
    bucket_counts[c] += 1
    if r["correct"]:
        bucket_correct[c] += 1

x_vals = []
y_vals = []

for c in ALLOWED_CONFIDENCE:
    if bucket_counts[c] > 0:
        x_vals.append(c / 100)
        y_vals.append(bucket_correct[c] / bucket_counts[c])

print("\nCalibration by confidence level:")
for c in ALLOWED_CONFIDENCE:
    if bucket_counts[c] > 0:
        acc = bucket_correct[c] / bucket_counts[c]
        print(f"{c}% confidence: {bucket_correct[c]}/{bucket_counts[c]} correct = {acc:.1%}")

# =========================================================
# PLOT
# =========================================================
# =========================================================
# BETTER CALIBRATION PLOT
# =========================================================
plt.figure(figsize=(8, 8))

# Convert to percentages
x_plot = [x * 100 for x in x_vals]
y_plot = [y * 100 for y in y_vals]

# Perfect calibration line
plt.plot([50, 100], [50, 100], linestyle="--", label="Perfect calibration")

# Observed calibration
plt.plot(x_plot, y_plot, marker="o", markersize=10, linewidth=2, label="Observed accuracy")

# Add labels for each point
for c in ALLOWED_CONFIDENCE:
    if bucket_counts[c] > 0:
        acc = bucket_correct[c] / bucket_counts[c] * 100
        plt.text(
            c,
            acc + 2,   # small vertical offset
            f"{c}% (n={bucket_counts[c]})",
            ha="center"
        )

plt.xlim(50, 100)
plt.ylim(0, 105)   # a little headroom above 100
plt.xticks(ALLOWED_CONFIDENCE)
plt.yticks(range(0, 101, 10))

plt.xlabel("Stated confidence (%)")
plt.ylabel("Observed accuracy (%)")
plt.title("Calibration Curve")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("NYS_for_Web_2025.csv")

# -----------------------------------
# 1. Check important columns
# -----------------------------------
key_cols = [
    "Release Decision at Arraign",
    "Remanded_to_Jail_at_Arraign",
    "Bail_Set_and_Posted_at_Arraign",
    "Bail_Set_and_Not_Posted_at_Arraign",
    "Gender",
    "Race",
    "Ethnicity",
    "County_Name",
    "Age_at_Arrest",
    "prior_vfo_cnt",
    "prior_nonvfo_cnt",
    "prior_misd_cnt",
    "pend_vfo",
    "pend_nonvfo",
    "pend_misd",
    "supervision"
]

print("=== CHECK WHETHER THESE COLUMNS EXIST ===")
for col in key_cols:
    print(f"{col}: {col in df.columns}")

# -----------------------------------
# 2. Inspect the main arraignment decision variable
# -----------------------------------
release_col = "Release Decision at Arraign"

print("\n=== RAW VALUE COUNTS: RELEASE DECISION AT ARRAIGN ===")
print(df[release_col].value_counts(dropna=False))

# Clean strings a bit
df[release_col] = df[release_col].astype("string").str.strip()

# -----------------------------------
# 3. Create a defensible binary outcome
# -----------------------------------
# 1 = released at arraignment
# 0 = not released at arraignment
# NaN = ambiguous / exclude from binary comparison
release_map = {
    "ROR": 1,
    "Nonmonetary release": 1,
    "Bail-set": 0,
    "Remanded": 0,
    "Disposed at arraign": np.nan,
    "Unknown": np.nan
}

df["released_clean"] = df[release_col].map(release_map)

print("\n=== CLEAN RELEASE VARIABLE (PROPORTIONS) ===")
print(df["released_clean"].value_counts(dropna=False, normalize=True))

print("\n=== CLEAN RELEASE VARIABLE (COUNTS) ===")
print(df["released_clean"].value_counts(dropna=False))

# Keep only rows with clear release/not-release status
analysis_df = df[df["released_clean"].notna()].copy()

print("\n=== ANALYSIS SAMPLE SIZE ===")
print(len(analysis_df))

# -----------------------------------
# 4. Group comparisons
# -----------------------------------
for var in ["Gender", "Race", "Ethnicity", "County_Name"]:
    if var in analysis_df.columns:
        print(f"\n=== RELEASE RATE BY {var} ===")
        rates = analysis_df.groupby(var)["released_clean"].mean().sort_values(ascending=False)
        counts = analysis_df[var].value_counts()

        result = pd.DataFrame({
            "release_rate": rates,
            "n": counts
        }).sort_values("release_rate", ascending=False)

        # For county, suppress tiny groups a bit
        if var == "County_Name":
            result = result[result["n"] >= 30]

        print(result.head(15))
        print("...")
        print(result.tail(15))

# -----------------------------------
# 5. Compare numeric variables by release status
# -----------------------------------
num_vars = [
    "Age_at_Arrest",
    "prior_vfo_cnt",
    "prior_nonvfo_cnt",
    "prior_misd_cnt",
    "pend_vfo",
    "pend_nonvfo",
    "pend_misd",
    "supervision"
]

available_num_vars = [col for col in num_vars if col in analysis_df.columns]

print("\n=== NUMERIC MEANS BY CLEAN RELEASE STATUS ===")
print(analysis_df.groupby("released_clean")[available_num_vars].mean())

# -----------------------------------
# 6. Correlations with release outcome
# -----------------------------------
print("\n=== CORRELATIONS WITH released_clean ===")
numeric_df = analysis_df.select_dtypes(include=[np.number]).copy()

# Avoid useless near-empty columns
valid_numeric_cols = numeric_df.columns[numeric_df.notna().mean() > 0.05]
numeric_df = numeric_df[valid_numeric_cols]

corrs = numeric_df.corr(numeric_only=True)["released_clean"].drop("released_clean").sort_values()

print("\nMost negative correlations:")
print(corrs.head(15))

print("\nMost positive correlations:")
print(corrs.tail(15))

# -----------------------------------
# 7. Simple cross-tab of the raw decision categories
# -----------------------------------
for var in ["Gender", "Race", "Ethnicity"]:
    if var in df.columns:
        print(f"\n=== RAW DECISION DISTRIBUTION BY {var} ===")
        tab = pd.crosstab(df[var], df[release_col], normalize="index")
        print(tab)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("NYS_for_Web_2025.csv")

decision = df["Release Decision at Arraign"].astype("string").str.strip()

decision_3cat = {
    "ROR": "Released",
    "Nonmonetary release": "Released",
    "Bail-set": "Bail/Detained",
    "Remanded": "Bail/Detained",
    "Disposed at arraign": "Disposed/Other",
    "Unknown": "Disposed/Other"
}

df["decision_3cat"] = decision.map(decision_3cat)

print("=== 3-CATEGORY OUTCOME DISTRIBUTION ===")
print(df["decision_3cat"].value_counts(normalize=True, dropna=False))

for var in ["Gender", "Race", "Ethnicity"]:
    print(f"\n=== {var} x 3-CATEGORY OUTCOME ===")
    tab = pd.crosstab(df[var], df["decision_3cat"], normalize="index")
    print(tab)
```
